```python
import os
import json
import time
import pandas as pd
import multiprocessing
from concurrent.futures import ProcessPoolExecutor, as_completed
from google import genai
from google.genai import types
import sys

# 1만 자리 이상의 숫자를 처리하기 위한 보안 설정
sys.set_int_max_str_digits(20000)

# ---------------------------------------------------------
# [설정값]
# ---------------------------------------------------------
ALL_KEYS = [os.environ.get(f"GEMINI_API_KEY_{i}") for i in range(1, 18)]

PROBLEMS_FILE = 'problems.csv'
TESTCASES_FILE = 'delete_testcases.csv'
OUTPUT_FILE = 'rangeValidated_testcases.csv'
TEMP_OUTPUT_FILE = OUTPUT_FILE + '.tmp'

BATCH_SIZE = 5
MAX_WORKERS = 13
MAX_RETRIES = 1
RETRY_DELAY = 5        # 재시도 전 대기 (초)
KEY_WAIT_TIMEOUT = 20  # 키 대기 최대 시간 (초)
KEY_COOLDOWN = 20      # 키 에러 후 선형 대기 기준 (초)

PROMPT_TEMPLATE = """
[Problem Description & Constraints]
{problem_context}

[Test Cases to Validate]
{testcases_batch_json}

[Task]
Return a JSON array consisting ONLY of boolean values (true or false).
- Return 'true' if the input perfectly satisfies ALL problem constraints.
- Return 'false' if the input violates ANY constraint.
- The length of the array must be exactly {batch_len}.

[Strict Output Requirements]
- You MUST return ONLY a JSON array.
- Do NOT include any explanations, markdown blocks, keys, or numbering.
Example output: [true, false, true, true]
"""

# ---------------------------------------------------------
# [워커 프로세스 함수]
# ---------------------------------------------------------
def worker_task(problem_context, testcases_batch, key_queue, key_error_times):
    """
    key_error_times: Manager().dict() — {api_key: last_error_unix_timestamp}
    """
    pid = os.getpid()
    start_time = time.time()

    prompt = PROMPT_TEMPLATE.format(
        problem_context=problem_context,
        testcases_batch_json=json.dumps(testcases_batch, ensure_ascii=False),
        batch_len=len(testcases_batch)
    )

    last_exception = None

    for attempt in range(1, MAX_RETRIES + 1):
        # --- 키 획득 (최대 KEY_WAIT_TIMEOUT 초 대기) ---
        key = None
        wait_start = time.time()
        while True:
            try:
                key = key_queue.get(timeout=1)
                break
            except Exception:
                if time.time() - wait_start >= KEY_WAIT_TIMEOUT:
                    return (
                        ["ERROR"] * len(testcases_batch),
                        time.time() - start_time,
                        pid,
                        {"true": 0, "false": 0, "ERROR": len(testcases_batch)},
                        "KEY_TIMEOUT",
                    )

        # --- 에러 키 쿨다운: 선형 대기 ---
        last_err = key_error_times.get(key)
        if last_err is not None:
            elapsed_since_err = time.time() - last_err
            cooldown = max(0.0, KEY_COOLDOWN - elapsed_since_err)
            if cooldown > 0:
                time.sleep(cooldown)

        # --- API 호출 ---
        try:
            client = genai.Client(api_key=key)
            response = client.models.generate_content(
                model='gemma-4-31b-it',
                contents=prompt,
                config=types.GenerateContentConfig(
                    response_mime_type="application/json",
                    temperature=0.1,
                ),
            )
            results = json.loads(response.text.strip())

            if not isinstance(results, list) or len(results) != len(testcases_batch):
                raise ValueError(f"Invalid response format: {response.text[:200]}")

            # 성공 — bool을 문자열로 통일하여 저장
            results_str = ["true" if r is True else "false" for r in results]
            key_queue.put(key)  # 키 반납

            counts = {"true": results_str.count("true"), "false": results_str.count("false"), "ERROR": 0}
            return results_str, time.time() - start_time, pid, counts, f"SUCCESS(attempt={attempt})"

        except Exception as e:
            last_exception = e
            # 에러 시간 기록 후 키 반납
            key_error_times[key] = time.time()
            key_queue.put(key)

            if attempt < MAX_RETRIES:
                time.sleep(RETRY_DELAY)

    # 3회 모두 실패
    counts = {"true": 0, "false": 0, "ERROR": len(testcases_batch)}
    return (
        ["ERROR"] * len(testcases_batch),
        time.time() - start_time,
        pid,
        counts,
        f"ERROR(all {MAX_RETRIES} attempts failed): {last_exception}",
    )


# ---------------------------------------------------------
# [메인 마스터 프로세스]
# ---------------------------------------------------------
if __name__ == "__main__":
    manager = multiprocessing.Manager()
    key_queue = manager.Queue()
    key_error_times = manager.dict()  # {api_key: last_error_unix_timestamp}

    for k in ALL_KEYS:
        key_queue.put(k)

    # 데이터 로드
    df_problems = pd.read_csv(PROBLEMS_FILE)
    if os.path.exists(OUTPUT_FILE):
        df_tc = pd.read_csv(OUTPUT_FILE)
    else:
        df_tc = pd.read_csv(TESTCASES_FILE)
        df_tc['rangeValidation'] = None

    df_problems['id'] = df_problems['id'].astype(str)
    df_tc['problem_id'] = df_tc['problem_id'].astype(str)

    unique_probs = df_tc['problem_id'].unique()
    total_unique_probs = len(unique_probs)

    print(f"🚀 총 {total_unique_probs}개의 문제에 대해 제약조건 검증을 시작합니다.")

    with ProcessPoolExecutor(max_workers=MAX_WORKERS) as executor:
        # future → (batch_index, prob_id) 매핑
        futures = {}

        for prob_idx, prob_id in enumerate(unique_probs, start=1):
            mask = df_tc['problem_id'] == prob_id

            # 이어하기: rangeValidation이 None이거나 "ERROR"인 행만 재처리
            '''
            target_rows = df_tc[
                mask & (
                    df_tc['rangeValidation'].isnull() |
                    (df_tc['rangeValidation'].astype(str) == 'ERROR')
                )
            ]
            '''

            # 이어하기 : rangeValidation이 None인 것만 재처리
            target_rows = df_tc[
                mask & (
                    df_tc['rangeValidation'].isnull()
                )
            ]

            if target_rows.empty:
                continue

            prob_text = df_problems[df_problems['id'] == prob_id]['question'].iloc[0]

            for i in range(0, len(target_rows), BATCH_SIZE):
                batch = target_rows.iloc[i: i + BATCH_SIZE]
                future = executor.submit(
                    worker_task,
                    prob_text,
                    batch.to_dict('records'),
                    key_queue,
                    key_error_times,
                )
                futures[future] = (batch.index, prob_id, prob_idx)

        completed = 0
        total_futures = len(futures)

        for future in as_completed(futures):
            indices, prob_id, prob_idx = futures[future]
            results, elapsed, pid, counts, status = future.result()

            df_tc.loc[indices, 'rangeValidation'] = results
            completed += 1

            print(
                f"[{completed}/{total_futures}] Problem {prob_idx}/{total_unique_probs} "
                f"(ID: {prob_id}) | PID:{pid} | Time:{elapsed:.2f}s | "
                f"T:{counts['true']} F:{counts['false']} E:{counts['ERROR']} | {status}"
            )

            # 배치 단위 안전 저장
            df_tc.to_csv(TEMP_OUTPUT_FILE, index=False, encoding='utf-8-sig')
            os.replace(TEMP_OUTPUT_FILE, OUTPUT_FILE)

    print("🎉 모든 작업 완료!")
```
