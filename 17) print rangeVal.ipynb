```python
import pandas as pd
from IPython.display import display, HTML

# ---------------------------------------------------------
# [설정값]
# ---------------------------------------------------------
# 검토하고 싶은 문제의 ID (문자열/숫자 무관하게 입력)
TARGET_PROBLEM_ID = "1953"  

PROBLEMS_FILE = 'problems - 복사본.csv'
TESTCASES_FILE = 'rangeValidated_testcases - 복사본.csv'

# ---------------------------------------------------------
# [데이터 로드 및 전처리]
# ---------------------------------------------------------
try:
    df_problems = pd.read_csv(PROBLEMS_FILE)
    df_testcases = pd.read_csv(TESTCASES_FILE)
except FileNotFoundError as e:
    print(f"❌ 파일을 찾을 수 없습니다: {e}")
    raise

# 안전한 매칭을 위해 ID를 순수 문자열로 변환 (소수점 찌꺼기 제거)
df_problems['id'] = df_problems['id'].astype(str).str.replace(r'\.0$', '', regex=True)
df_testcases['problem_id'] = df_testcases['problem_id'].astype(str).str.replace(r'\.0$', '', regex=True)

target_id_str = str(TARGET_PROBLEM_ID).replace(r'\.0$', '')

# display 헬퍼
def display_with_newlines(df, max_rows=15):
    return HTML(
        df.head(max_rows).to_html().replace(r'\n', '<br>')
    )

# ---------------------------------------------------------
# [1. 문제 지문(Question) 예쁘게 출력하기]
# ---------------------------------------------------------
target_prob = df_problems[df_problems['id'] == target_id_str]

if target_prob.empty:
    display(HTML(f"<h3 style='color: red;'>❌ 문제 ID {TARGET_PROBLEM_ID}을(를) {PROBLEMS_FILE}에서 찾을 수 없습니다.</h3>"))
else:
    question_text = target_prob.iloc[0]['question']
    
    # HTML과 CSS를 활용하여 가독성 높은 박스 형태로 렌더링
    box_html = f"""
    <div style="border: 2px solid #4CAF50; border-radius: 8px; padding: 20px; background-color: #fcfcfc; margin-bottom: 20px;">
        <h3 style="margin-top: 0; color: #2E7D32; border-bottom: 1px solid #4CAF50; padding-bottom: 10px;">
            📝 Problem {TARGET_PROBLEM_ID} - Question Description
        </h3>
        <pre style="white-space: pre-wrap; word-wrap: break-word; font-family: 'Malgun Gothic', 'Apple SD Gothic Neo', sans-serif; font-size: 15px; line-height: 1.6; margin: 0;">{question_text}</pre>
    </div>
    """
    display(HTML(box_html))

# ---------------------------------------------------------
# [2. 테스트케이스 True/False 분리 및 출력]
# ---------------------------------------------------------
target_tc = df_testcases[df_testcases['problem_id'] == target_id_str].copy()

if target_tc.empty:
    display(HTML(f"<h4 style='color: red;'>❌ 문제 ID {TARGET_PROBLEM_ID}에 해당하는 테스트케이스가 없습니다.</h4>"))
else:
    # boolean, 문자열('True', 'false') 등 섞여있을 수 있으므로 대문자 문자열로 완벽히 통일하여 검사
    val_str = target_tc['rangeValidation'].astype(str).str.strip().str.upper()
    
    df_true = target_tc[val_str == 'TRUE']
    df_false = target_tc[val_str == 'FALSE']
    df_error = target_tc[(val_str != 'TRUE') & (val_str != 'FALSE')]  # ERROR 나 NULL 인 경우
    
    # 출력에 방해되는 problem_id 열은 보기 좋게 숨김 처리
    display_cols = [col for col in target_tc.columns if col != 'problem_id']

    # --- TRUE 테이블 출력 ---
    display(HTML(f"<h3 style='color: #1565C0; margin-top: 20px;'>✅ 제약조건 충족 (True) - 총 {len(df_true)}개</h3>"))
    if not df_true.empty:
        display(display_with_newlines(df_true[display_cols]))
    else:
        print("충족하는 테스트케이스가 없습니다.")
    
    # --- FALSE 테이블 출력 ---
    display(HTML(f"<h3 style='color: #c62828; margin-top: 30px;'>❌ 제약조건 위반 (False) - 총 {len(df_false)}개</h3>"))
    if not df_false.empty:
        display(display_with_newlines(df_false[display_cols]))
    else:
        print("위반하는 테스트케이스가 없습니다.")
    
    # --- ERROR 테이블 출력 ---
    if not df_error.empty:
        display(HTML(f"<h3 style='color: #e65100; margin-top: 30px;'>⚠️ 검증 에러/누락 (ERROR/NULL) - 총 {len(df_error)}개</h3>"))
        display(display_with_newlines(df_error[display_cols]))
```
